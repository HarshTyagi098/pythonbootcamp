# Loops
for loop and while loop

```python
### while loop
i=0
while i<5:
  print(i)
  i+=1
```

```python
i=0
while i<11:
  if(i%2==0):
    print(i)
  i+=1
```

```python
### for loop
##in this we use range function
for i in range(5):
  print(i)
```

```python
for i in range(1,11,1):   ### in range function (Start,Stop,Step size)
  print(i)
```

```python
### print odd number using for loop
for i in range(1,11,2):
  print(i)
print()
### print even number using for loop
for i in range(2,11,2):
  print(i)
```
