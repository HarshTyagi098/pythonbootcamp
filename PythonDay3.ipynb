#Now take new dataframe
to fill null values

```python
df3=pd.DataFrame({
    'Name':['ram','mohan','rohit'],
    'Rollno':[None,2,3],
    'Marks':[45,56,None]
})
df3
```

```python
df3['Rollno']=df3['Rollno'].fillna(1)       # here we fill roll number with 1 as per our IQ
df3
```

```python
df3['Marks'].mean()          # here we find mean of marks column
```

```python
df3['Marks']=df3['Marks'].fillna(df3['Marks'].mean())       # here we fill marks with mean as per our IQ
df3
```

```python
df4=pd.DataFrame({
    'Name':['ram','mohan','rohit','sonal'],
    'Rollno':[1,2,3,4],
    'Marks':[45,56,67,89],
    'Gender':['M',None,'M','F']
})
df4
```

```python
df4.isnull().sum()
```

```python
res=df4['Gender'].mode()[0]
res
```

```python
df4['Gender']=df4['Gender'].fillna(res)         # here we fill Gender with mode of gender column input
df4
```

#Histogram

```python
x=np.random.normal(170,10,250)    ## 170 is average mean, 10 is standard daviation, 250 is total data points
plt.hist(x)
plt.show()

#bell shape curve is it's property (Mean = Median = Mode)
```

#Pie chart

```python
x=np.array([35,25,25,15])
mylabels=["Apples","Bananas","Cherries","Dates"]

plt.pie(x,labels=mylabels)
plt.show()
```

#Start Angle
from where we want to start

```python
x=np.array([35,25,25,15])
mylabels=["Apples","Bananas","Cherries","Dates"]

plt.pie(x,labels=mylabels,startangle=90)
plt.show()
```

```python
# Adding percentage in Pie Chart
x=np.array([35,25,25,15])
mylabels=["Apples","Bananas","Cherries","Dates"]

plt.pie(x,labels=mylabels,startangle=90,autopct="%1.2f%%")
plt.show()
```

#Explode

```python
x=np.array([35,25,25,15])
mylabels=["Apples","Bananas","Cherries","Dates"]
myexplode=[0.2,0,0,0]
plt.pie(x,labels=mylabels,startangle=90,autopct="%1.2f%%",explode=myexplode)
plt.show()
```
