```python
import keras
print(keras.__version__)
```

we use "Sequential Class" code for model, we can costomised this model

```python
model = keras.Sequential()            #input layer
model.add(keras.Input(shape=(13,)))     # we will add layer one by one,   Independent Features
model.add(keras.layers.Dense(8,activation='relu'))    # hidden layer , only write in 2's power
model.add(keras.layers.Dense(8,activation='relu'))
model.add(keras.layers.Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])          # now we compile the model
```

copy the code of "Model traning api" code from Keras documentation

```python
model.fit(
    x=X_train,
    y=y_train,
    batch_size=32,
    epochs=50,
    verbose="auto",
    validation_data=(X_test,y_test),
)                                                  # After every Epoch accuracy increases, because model can improve on every Epoch
```

```python
model.evaluate(x=X_test,y=y_test,batch_size=32,verbose='auto')
```
