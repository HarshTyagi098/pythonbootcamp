```python
import nltk
nltk.download('punkt_tab')  # Tokenization     (convert sentence into words)
nltk.download('averaged_perceptron_tagger')  # POS Tagging   (noun,pronoun, adjective)
nltk.download('stopwords')  # Stop words     (a,an,the)
nltk.download('wordnet')  # Lemmatization
nltk.download('vader_lexicon')  # Sentiment Analysis
```

```python
#either we go with the sentence tokenizer or go with the word tokenizer     but not go with both
from nltk.tokenize import word_tokenize  # which word start when and where it stop
nltk.download('punkt_tab')
text='This is an example sentence'
tokens=word_tokenize(text)
print(tokens)
```

```python
from nltk.tokenize import sent_tokenize
text="Hello! How are you? This is an example."
sentences=sent_tokenize(text)
print(sentences)
```

removing the stop words or healping words

```python
from nltk.corpus import stopwords     # To remove the stop words like : is, a, this,the etc.
nltk.download('stopwords')
stop_words=set(stopwords.words('english'))
tokens=word_tokenize("This is a sample sentence, showing off the stop words filtration.")
filtered_tokens=[word for word in tokens if word.lower() not in stop_words]
print(filtered_tokens)
```

```python
from nltk.stem import PorterStemmer      # To convert into the root word from given word
ps=PorterStemmer()
words=["running","ran","runs"]
stemmed_words=[ps.stem(word) for word in words]
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

words = ["running", "ran", "runs"]
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]

print(lemmatized_words)
```

```python
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger_eng')
tokens=word_tokenize("This is a sample sentence.")
pos_tags=pos_tag(tokens)
print(pos_tags)
```

```python
import nltk
nltk.download('vader_lexicon')

from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Initialize the VADER sentiment analyzer
sid = SentimentIntensityAnalyzer()

# Sentence to analyze
sentence = "I am a great, positive, Briliant, peaceful, kind!"

# Perform sentiment analysis on the sentence
scores = sid.polarity_scores(sentence)

print(scores)
```

```python
from nltk import  ne_chunk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

# Tokenize the text into words
words = word_tokenize("Barack Obama was born in Hawaii. and his salary is 5 billion dollars")

# Perform POS tagging
pos_tags = pos_tag(words)

# Perform NER using ne_chunk
named_entities = ne_chunk(pos_tags)

print(named_entities)
```

```python
from nltk.probability import FreqDist
from nltk.tokenize import word_tokenize

text = "I love natural language processing. Processing is fun."
words = word_tokenize(text)
fdist = FreqDist(words)

print(words)
print(fdist)
```
