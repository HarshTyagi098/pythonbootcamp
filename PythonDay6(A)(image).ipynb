```python
x_train = x_train/255.0
```

```python
model = keras.Sequential()            ## This is an eempty model, we have to add the data by ownself
```

```python
model.summary()
```

now copy the code from keras Conv2D layer         (Convonational layer)
it extract features from images like (eyes, nose,etc)

```python
model.add(
    keras.layers.Conv2D(32,(3,3),
    activation='relu',
    input_shape=(32,32,3)
))
```

```python
model.add(             # to increase the accuracy
    keras.layers.Conv2D(64,(3,3),
    activation='relu',
))
```

Now adding maxpooling2D (Maxpool layer)

```python
model.add(keras.layers.MaxPooling2D(
    pool_size=(2, 2)
)
)
```

```python
model.add(keras.layers.MaxPooling2D(
    pool_size=(2, 2)
)
)
```

Now adding Flatten layer in our model (it will convert into single dimension)

```python
model.add(keras.layers.Flatten())
```

Now adding Dense layer in our model

```python
model.add(
    keras.layers.Dense(
    64,
    activation='relu'
))
```

now output layer that is softmax

```python
model.add(
        keras.layers.Dense(
    10,
    activation='softmax'
)
)
```

```python
model.summary()
```

picking code from https://keras.io/api/models/model_training_apis/

```python
model.compile(
    optimizer="adam",
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
callback=keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=3,
    verbose=0,
    restore_best_weights=True,
)              ## it will increase accuracy
```

When we think accuracy not increase and we want to stop the Epoch, that's why we use Call back(Early stoping)
so early stoping stop it by ownself after checking

https://keras.io/api/callbacks/early_stopping/

```python
model.fit(
    x=x_train,
    y=y_train,
    batch_size=64,
    epochs=100,
    verbose="auto",
    validation_data=(x_test,y_test),
    callbacks=[callback]
)
```

```python
model.evaluate(
    x=x_test,
    y=y_test,
    batch_size=32,
    verbose="auto",
)
```

# increase accuracy
now we add one more convonational and maxpool layer to increase the accuracy

```python
model.predict(x_test, batch_size=32, verbose="auto")
```
