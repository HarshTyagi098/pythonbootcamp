# Mnist digit classification dataset
use data set from https://keras.io/api/datasets/mnist/

hand written digits from 0 to 9

```python
import keras
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

for channel we have to reshape

```python
x_train = x_train.reshape(-1, 28, 28, 1)       # only change in x because it is independent , and y is dependent
x_test = x_test.reshape(-1, 28, 28, 1)
```

```python
x_train.shape                   # here 1 means in channel is grey color
```

```python
x_test.shape
```

```python
x_train = x_train/255.0
```

```python
x_test = x_test/255.0
```

# Now creating model

```python
model = keras.Sequential()
```

```python
model.summary()
```

```python
model.add(
    keras.layers.Conv2D(32,(3,3),
    activation='relu',
    input_shape=(28,28,1)
))
```

```python
model.add(keras.layers.MaxPooling2D(
    pool_size=(2, 2)
)
)
```

```python
model.add(             # to increase the accuracy
    keras.layers.Conv2D(64,(3,3),
    activation='relu',
))

model.add(keras.layers.MaxPooling2D(
    pool_size=(2, 2)
)
)
```

```python
model.add(             # to increase the accuracy
    keras.layers.Conv2D(64,(3,3),
    activation='relu',
))

model.add(keras.layers.MaxPooling2D(
    pool_size=(2, 2)
)
)
```

```python
model.add(keras.layers.Flatten())
```

```python
model.add(
    keras.layers.Dense(
    64,
    activation='relu'
))
```

```python
model.add(
        keras.layers.Dense(
    10,
    activation='softmax'
)
)
```

```python
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
callback=keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=3,
    verbose=0,
    restore_best_weights=True,
)              ## it will increase accuracy
```

```python
model.fit(
    x=x_train,
    y=y_train,
    batch_size=64,
    epochs=100,
    verbose="auto",
    validation_data=(x_test,y_test),
    callbacks=[callback]
)
```

```python
model.evaluate(
    x=x_test,
    y=y_test,
    batch_size=32,
    verbose="auto",
)
```
