# Iris Dataset ANN (Deep Learning)

```python
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df['species'] = le.fit_transform(df['species'])        # now convert object type into numeric type
df
```

```python
df.info()
```

```python
X=df.drop('species',axis=1)      # axis 1 means we are working on columns, axis 2 means we are working on rows
y=df['species']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
import keras
```

```python
model = keras.Sequential()
```

```python
model.summary()
```

```python
model.add(
    keras.Input(shape=(4,),
    batch_size=32
)
)
```

```python
model.add(keras.layers.Dense(10,activation='relu'))
model.add(keras.layers.Dense(3,activation='softmax'))
```

```python
model.summary
```

```python
model.compile(
    optimizer="adam",
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
callback=keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=3,
    verbose=0,
    restore_best_weights=True,
)              ## it will increase accuracy
```

```python
model.fit(
    x=X_train,
    y=y_train,
    batch_size=64,
    epochs=100,
    verbose="auto",
    validation_data=(X_test,y_test),
    callbacks=[callback]
)
```

```python
model.evaluate(
    x=X_test,
    y=y_test,
    batch_size=32,
    verbose="auto",
)
```
