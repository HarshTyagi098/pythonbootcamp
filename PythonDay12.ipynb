```python
# difference between opening and clossing
df['open-close'] = df['Open'] - df['Close']
df['low-high'] = df['Low'] - df['High']
df['target'] = np.where(df['Close'].shift(-1) > df['Close'],1,0)
```

# Deap Learning

```python
import keras
```

```python
model = keras.Sequential()
```

```python
model.summary()
```

```python
model.add(
    keras.Input(
        shape=(3,),
        batch_size=32
    )
)
```

```python
model.add(                                 #hidden layer
    keras.layers.Dense(
    64,
    activation='relu'
))
```

```python
model.add(                      # hidden layer
    keras.layers.Dense(
    128,
    activation='relu'
))
```

```python
model.add(                      # hidden layer
    keras.layers.Dense(
    256,
    activation='relu'
))
```

```python
model.add(                      # hidden layer
    keras.layers.Dense(
    512,
    activation='relu'
))
```

```python
model.add(                         ## Output layer
        keras.layers.Dense(
    1,
    activation='sigmoid'   # use when we check 1 and 0
)
)
```

```python
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
from gc import callbacks
callback=keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=5,
    restore_best_weights=True
)
```

```python
model.fit(
    x=X_train,
    y=y_train,
    batch_size=32,
    epochs=100,
    verbose="auto",
    validation_data=(X_test,y_test),
    callbacks=[callback]
)
```

```python
model.evaluate(X_test,y_test)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Drop non-numeric columns, if any (e.g., Date)
df_numeric = df.select_dtypes(include=['float64', 'int64'])

# Standardize the data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_numeric)

# Apply PCA
pca = PCA(n_components=2)  # You can adjust the number of components
principal_components = pca.fit_transform(df_scaled)

# Create a DataFrame with principal components
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])

# Optional: Visualize the PCA result
plt.figure(figsize=(8, 6))
sns.scatterplot(x='PC1', y='PC2', data=pca_df)
plt.title('PCA of Tesla Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error
```

```python
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 5, 10]
}
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(random_state=42)
```

```python
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=5, n_jobs=-1, scoring='neg_mean_squared_error')
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("Best Score:", -grid_search.best_score_)

# Evaluate on test set
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
print("Test MSE:", mse)
```
