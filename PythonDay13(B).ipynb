```python
from math import pi
from transformers import pipeline

summarizer = pipeline("summarization")
classifier = pipeline("sentiment-analysis")
qa = pipeline("question-answering")
translator = pipeline("translation_en_to_fr")
ner = pipeline("ner", grouped_entities=True)
```

```python
print(summarizer("when was India Independent",))
print(classifier("when was India Independent"))
print(qa(question="when was India Independent", context="when was India Independent"))
print(translator("when was India Independent",))
print(ner("when was India Independent"))
```
